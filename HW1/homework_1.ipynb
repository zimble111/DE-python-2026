## Задание 1

```python
import json

purchases = {}

with open('purchase_log.txt', encoding='utf-8') as file:
    next(file)
    for line in file:
        purchase = json.loads(line)
        purchases[purchase['user_id']] = purchase['category']

for user_id in list(purchases)[:2]:
    print(user_id, repr(purchases[user_id]))
```

## Задание 2

```python
import csv

with open('visit_log.csv', encoding='utf-8', newline='') as visits, \
     open('funnel.csv', 'w', encoding='utf-8', newline='') as funnel:
    reader = csv.reader(visits)
    writer = csv.writer(funnel, lineterminator='\n')
    next(reader)
    writer.writerow(['user_id', 'source', 'category'])

    for user_id, source in reader:
        if user_id in purchases:
            writer.writerow([user_id, source, purchases[user_id]])

with open('funnel.csv', encoding='utf-8') as file:
    for _ in range(3):
        print(file.readline().strip())
```
