## Задание 1

```python
from datetime import datetime
import locale

locale.setlocale(locale.LC_TIME, 'C')

newspapers = [
    ('Wednesday, October 2, 2002', '%A, %B %d, %Y'),
    ('Friday, 11.10.13', '%A, %d.%m.%y'),
    ('Thursday, 18 August 1977', '%A, %d %B %Y')
]

for text, date_format in newspapers:
    print(datetime.strptime(text, date_format))
```

## Задание 2

```python
from datetime import timedelta


def date_range(start_date, end_date):
    try:
        start = datetime.strptime(start_date, '%Y-%m-%d').date()
        end = datetime.strptime(end_date, '%Y-%m-%d').date()
    except (ValueError, TypeError):
        return []

    if start.isoformat() != start_date or end.isoformat() != end_date:
        return []

    if start > end:
        return []

    return [
        (start + timedelta(days=i)).isoformat()
        for i in range((end - start).days + 1)
    ]


print(date_range('2022-01-01', '2022-01-03'))
print(date_range('2022-01-03', '2022-01-01'))
print(date_range('2022-02-30', '2022-02-31'))
```
